### 5. Le slicing de séquences

A partir d'une séquence donnée, on veut extraire une sous-séquence. Cela fonctionne de manière similaire à un intervalle mathématiques: séquence[début:fin[. 

La syntaxe générale est L[i:j:k]: 
- i = indice du début
- j = indice de fin 
- k = le "pas" (par défaut il vaut 1)

La sous-séquence sera donc composée de tous les éléments de l'indice i jusqu'à l'indice j-1 par pas de k. Cette sous-séquence est un nouvel objet du même type que la séquence initiale. 

```python
exemple_list = [1, 2, 3, 4, 5]
print("Le type de ma séquence est:", type(exemple_list))

sous_exemple_list = exemple_list[0:4:2]
print("Ma sous séquence est:", sous_exemple_list)
print("Le type de ma sous séquence est:", type(sous_exemple_list))
```

```python
# Quelques exemples 

L = 'Dans le Python tout est bon'.split()
print("La liste L est:", L)
print("Le type de L est:", type(L))

# Sous liste contenant les 3 premiers éléments de L
print(L[0:3:1], "ou", L[0:3], "ou, encore", L[:3])

# Sous liste contenant les 3 derniers éléments de L
print(L[3:len(L):1], "ou", L[3::1], "ou, encore", L[3:])

# Sous liste excluant le dernier élément
print(L[0:-1:1], "ou", L[:-1:1], "ou, encore", L[:-1])
```

*Remarque*: ici *split( )* est une méthode associée à l'objet chaînes de caractères qui permet de découper une chaîne de caractères en plusieurs morceaux dans une liste.

### 6. Boucles

Les boucles sont des structures qui permettents de répéter l'éxécution d'un bloc de code plusieurs fois.

#### Boucles for 

On peut notamment utiliser les boucles for pour manipuler des listes d'élements. Par exemple, le code ci-dessous parcours la liste $[2,8,5]$ élément par élement pour afficher l'élement et son carré. 

```python
for i in [2, 8, 5]: 
    print(i, i * i)
```

Que fait-le code suivant ? 

```python
u = [2, 8, 5]
for i in u: 
    print(i, i * i)
```

On peut aussi itérer sur des chaînes de caractères ! 

```python
invites = ['Pierre', 'Mathilde', 'Axel', 'Morgane', 'Bart']
for invite in invites: 
    print(f'Bonjour {invite}, bienvenue à la maison !')
print("Maintenant, tout le monde est arrivé.")
```

Et même sur des dictionnaires ! 

```python
invites = {'Pierre': False, 'Mathilde': True, 'Axel': True, 'Morgane': False, 'Bart': True}

for (personne, presence) in invites.items(): 
    if presence: 
        print(f'Bonjour {personne}, bienvenue à la maison !')
    else: 
        print(f'Dommage, {personne} n\'est pas là !')
```

*Remarque*: *item( )* est une méthode liée aux dictionnaires pour renvoyer une vue itérable des éléments du dictionnaires. 



### Boucles while 

Tant que la condition est True alors le code est exécuté, dans la condition passe à False l'exécution est arrêtée. 



```python
compteur = 22
k = 0 
while compteur > 0: 
    print('le compteur vaut:', compteur)
    compteur -= 1
    k += 1
print('le compteur a été décrémenté', k, 'fois et vaut maintenant', compteur)
```

Contrairement aux boucles for, la fonction n'a pas un nombre d'itération prédéfini, il faut donc faire attention aux boucles infinies! 

```python
# Exemple de boucle infinie 
# while True: 
#    print(1)
```

On peut arrêter l'exécution d'une boucle avant la fin grâce à l'instruction break. 

*Remarque*: on peut aussi l'utiliser dans une boucle for

```python
compteur = 3 
while True: 
    compteur -= 1
    print("Dans la boucle infinie ! compteur =", compteur)
    if compteur <= 0: 
        break

print("Ce n'était pas vraiment une boucle infinie.")
```

Dans le cas où plusieurs boucles sont imbriquée, l'instruction break sort de la plus imbriquée. 

```python
for i in (1, 2, 3): 
    for j in (1, 2, 3, 4):
        if i == 2: 
            break 
        print(f'i = {i}, j = {j}')
```

Si, dans une boucle, on veut passer immédiation à l'itération suivante, on utilise l'instruction continue. 

```python
compteur = 9 
while compteur > 0: 
    compteur -= 1
    if compteur % 2 : 
        compteur /= 2
        print('non pair, on divise: ', compteur)
        continue
    print("pair, RAS")
```
