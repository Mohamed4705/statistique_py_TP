### 1.1 Créer un ndarray

Il existe trois méthodes pour créer un **ndarray**:
* à partir d'une liste; 
* grâce aux fonctions Numpy; 
* à partir d'un fichier. 

```python
# Creation à partir de liste
exemple_list1 = [1,2,3,4,5]
exemple_ndarray = np.array(exemple_list1)
#exemple_ndarray = numpy.array(exemple_list)
print("Le ndarray crée est: ", exemple_ndarray, "et son type est:", type(exemple_ndarray))

# Création d'un ndarray bidimensionnel à partir de listes
exemple_list2 = [5,4,6,7,8]
exemple_ndarray = np.array([exemple_list1, exemple_list2])
print("Le ndarray crée est: \n", exemple_ndarray, "\n et son type est:", type(exemple_ndarray))
```

```python
# Creation à partir des fonctions NumPy 
mon_array = np.zeros((4,6))
print("Cette fonction permet de créer un ndarray rempli de 0 de 4 lignes et 6 colonnes \n", mon_array)

mon_array = np.ones((3,5), dtype = float)
print("Cette fonction permet de créer un ndarray rempli de 1 de 3 lignes et 5 colonnes \n", mon_array)

mon_array = np.empty((4,3), dtype = int)
print("Cette fonction permet de créer un ndarray rempli d'élément aléatoire de type 'interger' de 3 lignes et 5 colonnes \n", mon_array)
```

On peut lire un fichier .csv ou .txt dans un ndarray. Attention, les ndarray ne doivent contenir qu'un seul type de données !

### 1.2 Indexation 

Comme pour les listes il est possible d'accéder à des éléments précis du **ndarray**. 

```python
# J'accède à la 3ème ligne du tableau 
print(iris[2])

# J'accède à la 3ème ligne / 2ème colonne du tableau 
print(iris[2][1])

# Je remplace la valeur dans le tableauprint(iris[2][1])
iris[2][1] = 3.5
print(iris[2])
```

On peut également sélectionner des éléments qui correspondent à des conditions. 

Par exemple, la commande ci dessous sélectionne tous les éléments supérieurs à 3 indépendamment des lignes ou des colonnes.

Enfin, on peut également utiliser le slicing, comme pour les listes, afin de sélectionner un **sous-ndarray**. 

```python
print(iris[:,0])
```

```python
print("Récupère les 2 premières lignes et 4 première colonnes \n", iris[:2,:4])

print("Récupère les lignes 2 et 4 et 4 première colonnes \n", iris[1:5:2,:4 ])

print("Récupère les 5 dernières lignes et 4 dernières colonnes \n", iris[145:,2: ])
```

*Remarque*: lorsque que l'on créer un sous-tableau grâce à un slicing on crée une *vue* du tableau et non une *copie*, c'est à dire que tout ce que l'on fait au sous tableau se répercute au vrai ! 

```python
# Création d'un sous tableau 
print("tableau: \n", iris[145:,2: ])
sub_iris = iris[145:,2: ]
print("sous-tableau: \n", sub_iris)

# Changement du sous-tableau 
sub_iris[1,1] = 10
print("sous-tableau: \n", sub_iris)

print("Le tableau a aussi changé ! \n", iris[145:,2: ])
```

Si vous voulez travailler sur un sous tableau sans en changer la valeur, vous devez utiliser la méthode *copy( )*

```python
# Création d'un sous tableau 
print("tableau: \n", iris[145:,2: ])
sub_iris = iris[145:,2: ].copy()
print("sous-tableau: \n", sub_iris)

# Changement du sous-tableau 
sub_iris[1,1] = 2
print("sous-tableau: \n", sub_iris)

print("Le tableau n'a pas changé : \n", iris[145:,2: ])
```

### 2.2 Nettoyer un dataframe 

En général la première chose à faire est d'explorer les données manquantes. Y en a-t-il ? Combien ? Pourquoi ? Veut-on les supprimer ? Les remplacer ? La méthode *isna()* permet de voir s'il y a des données manquantes. 

```python
# Y a-t-il des NAs ?
athlete.isna()
```

```python
# Combien ?
athlete.isna().sum()
```

On remarque ici qu'on a la somme des NAs sur chaque colonne ! Très pratique. 
On peut vouloir les supprimer: 

```python
athlete_wtt_na = athlete.dropna(how="any")
athlete_wtt_na.isna().sum()
```


Pour la variable *Medal*, il y a beaucoup de valeurs manquantes. C'est normal, cette variable considère 3 modalités: Or, Argent, Bronze, si l'athlète n'a pas eu de médailles alors ses performances ne sont pas enregistrées. 

En revanche, les variables âge, taille et poids ne sont pas toujours renseignés et cela semble être de véritables valeurs manquantes. 
On peut donc choisir les colonnes qui nous intéressent. 

```python
athlete_wtt_na = athlete[athlete["Height"].notna()].copy()
athlete_wtt_na.isna().sum()
```

On n'a donc plus de valeur manquante pour la variable Taille parce que les lignes ont été supprimées. On voit d'ailleurs que nombre de valeurs manquantes dans les autres colonnes diminuent. 

Le code ci-dessous est donné à titre d'exemple, dans le cas où on aurait voulu supprimer toutes les valeurs NA dans les variables taille, poids et âge. 

```python
athlete_wtt_na = athlete[athlete[["Height", "Weight", "Age"]].notna().all(axis=1)]
athlete_wtt_na.isna().sum()
```

Maintenant, dans certains cas, le fait de supprimer autant de valeur manquantes peut nous faire perdre de précieuse données. Donc, il existe des méthodes pour remplacer ces valeurs manquantes. Les plus basiques consistent à utiliser la moyenne (ou la médiane, ou le mode dans le cas de données catégorielle). 

```python
athlete_wtt_na[["Weight", "Age"]] = athlete_wtt_na[["Weight", "Age"]].fillna(athlete_wtt_na[["Weight", "Age"]].mean())
athlete_wtt_na['Medal'] = athlete_wtt_na['Medal'].fillna("No Medal")
athlete_wtt_na.isna().sum()
```

*Remarque*: Il existe d'autres méthodes pour gérer l'**imputation des données manquantes**. Elles ne seront pas abordés dans ce cours mais si vous devez vous lancer dans une étude statistique renseignez-vous ! 

Il peut aussi y avoir des problèmes liés à la duplication des données. Regardons si dans notre jeu de données, 1 ligne apparait plusieurs fois. 

```python
athlete_wtt_na[athlete_wtt_na.duplicated()]
```

On voit qu'il y a un certains nombre de ligne duppliquées. 

```python
athlete_wtt_na[athlete_wtt_na.duplicated()].shape
```

On va les supprimer car elles peuvent fausser les analyses. 

```python
print(athlete_wtt_na.shape)
athlete_wtt_na.drop_duplicates(inplace=True)
print(athlete_wtt_na.shape)
```
